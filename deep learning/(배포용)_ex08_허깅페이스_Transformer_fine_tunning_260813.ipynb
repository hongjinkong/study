# Transformer (BERT (Bidirectional Encoder Representations from Transformers))

### 시퀀스-투-시퀀스(Sequence-to-Sequence)
  - 번역기에서 대표적으로 사용되는 모델
  - 인코더와 디코더라는 두 개의 모듈로 구성
  - 인코더 :  입력 문장의 모든 단어들을 순차적으로 입력받은 뒤에 마지막에 이 모든 단어 정보들을 압축해서 하나의 벡터로 만듬 (context vector)
  - 디코더 : 컨텍스트 벡터를 받아서 번역된 단어를 한 개씩 순차적으로 출력

- RNN에 기반한 seq2seq 모델의 두 가지 문제
  - 하나의 고정된 크기의 벡터에 모든 정보를 압축하려고 하니까 정보 손실이 발생
  - RNN의 고질적인 문제인 기울기 소실(vanishing gradient) 문제가 존재

### Transformer
  - 필요한 것은 주의 집중(Attention is All You Need) 논문으로 2017년 발표
  - 순환과 컨볼루션을 완전히 배제하고 오로지 Attention 메커니즘에만 기반을 둔 새롭고 단순한 신경망 아키텍처

  - 인코더와 디코더로 구성
  - 인코더
    - 트랜스포머는 하이퍼파라미터인 num_layers 개수의 인코더 층을 쌓음
    - 하나의 인코더 층은 크게 셀프 어텐션과 피드 포워드 신경망의 2개의 서브층(sublayer)으로 구성
      - Multi-head Self-Attention : 셀프 어텐션을 병렬적으로 사용
      - Position-wise FFNN : 일반적인 피드 포워드 신경망
    - 인코딩의 출력은 디코더로 전달

  - 디코더
    - num_layers 개수의 디코더 층을 쌓음  
    - 인코더에서 보낸 출력을 디코더 층의 연산에 사용
    - 디코더는 3개 층으로 구성
      - Masked Multi-head Self-Attention (Self Attention과 Look-ahead mask): 현재 시점의 예측에서 현재 시점보다 미래에 있는 단어들을 참고하지 못하도록 룩-어헤드 마스크(look-ahead mask)를 사용
        - 트랜스포머는 문장 행렬로 입력을 한 번에 받으므로 현재 시점의 단어를 예측하고자 할 때, 입력 문장 행렬로부터 미래 시점의 단어까지도 참고할 수 있는 현상이 발생
      - Multi-head Self-Attention (Encoder-Decoder Attention) : 멀티 헤드 어텐션을 수행
      - Position-wise FFNN

- Masked Multi-head Attension
  - 학습과정에서는 인코더에 "어제 카페 갔다 거기 사람 많다"가, 디코더에 입력된 상황이라면 트랜스포머 모델은 다음 영어 단어 I를 맞추도록 학습
  - 하지만 학습 과정에서 모델에 이번에 맞춰야할 정답인 I를 알려주게 되면 학습하는 의미가 없어짐
  - 따라서 정답을 포함한 미래 정보를 셀프 어텐션 계산에서 제외하게 하는 것
  - 예측할 대상은 확률은 높이고 다른 단어들의 확률은 낮게 조정

### BERT
- BERT 모형은 2018년 11월 구글이 공개한 인공지능(AI) 언어 모델
- BERT 모형의 특징 : 사전학습, 문맥학습, 파인튜닝
  - (사전학습) 위키피디아 같은 아주 큰 데이터들을 사용하여 '언어 이해' 모델을 사전학습(Pre-training)
  - (문맥학습1) 문장 순서를 학습하여 다음에 나온 문장이 순서에 맞는 문장인지 학습
    - 문장1: 저 남자는 회사에 출근했다
    - 문장2: 회사에 출근하자마자 저 남자는 커피를 끓여 마셨다. (순서가 맞음)
    - 문장3: 저 여자는 퇴근하려 한다.
    - 문장4: 강아지는 예쁘다. (순서가 틀린 문장)

  - (문맥학습2) 양방향으로 학습하여 가려진 단어를 맞춘다.
    - 문장 : 저 남자는 (①)에 출근했다. 회사에 출근하자마자 저 남자는 (②)를 마셨다.
    - ① : 회사 ② : 커피

  - (파인튜닝) 생성된 모델을 최적화하여 새로운 과제를 해결

- 데이터 새로 불러오기

```python
# 네이버영화리뷰 데이터 불러오기
import pandas as pd

train_data = pd.read_csv('./data/ratings_train.txt', delimiter='\t')
test_data  = pd.read_csv('./data/ratings_test.txt', delimiter='\t')

# 결측치 제거
train_data = train_data.dropna(how='any')
test_data = test_data.dropna(how='any')

# 한글과 공백만 추출, 댓글이 없는 데이터 삭제
train_data["document"] = train_data["document"].str.replace("[^가-힣 ]","", regex=True)
train_data = train_data[train_data['document'].str.strip() != '']

test_data["document"] = test_data["document"].str.replace("[^가-힣 ]","", regex=True)
test_data = test_data[test_data['document'].str.strip() != '']
```

- 일부 데이터만 사용

```python
import pandas as pd
# 훈련데이터 15만개 -> 다학습시키려면 시간이 오래 걸림 -> 앞, 뒤 1만개씩 2만개 데이터만 사용
train_data2 = pd.concat([train_data.iloc[:10000],
                         train_data.iloc[-10000:]])# 마이너스(-) 기호는 뒤에서부터 센다는 뜻/데이터셋의 맨 마지막부터 역순으로 뒤의 10,000개 행을 가져옴
train_data2
```

- 훈련-검증 데이터 분할

```python
from sklearn.model_selection import train_test_split

# X와 y를 따로 넣지 않고, train_data2라는 데이터프레임(통째로 된 표 하나)을 그대로 집어 넣음
# 표 한 장을 행(Row) 기준으로 8:2로 쪼갬 ➡️ 총 2개 반환 (train_df, val_df)
train_df, val_df = train_test_split(train_data2,
                                    test_size=0.2,# df 하나 넣음 -> 8:2로 쪼개진 2개의 리스트 나옴
                                    random_state=42)

# 문자열 데이터로 명시 후 리스트로 변환
# 허깅페이스 모델이 알아들을 수 있는 형태로 변환해야해서 리스트로 바꾸는 중!
train_texts = train_df['document'].astype(str).tolist()
train_labels = train_df['label'].tolist()
val_texts = val_df['document'].astype(str).tolist()
val_labels = val_df['label'].tolist()
```

# KoBERT
- 2019년에 SKT Brain에서 공개
- 한국어 자연어 처리를 위해 최적화된 BERT(Bidirectional Encoder Representations from Transformers) 기반 모델
- KoBERT는 BERT 모델을 한국어 데이터로 사전 학습하여, 한국어의 미묘한 문맥과 의미를 더 잘 파악할 수 있도록 만들어진 라이브러리

### Hugging Face Fine Tuning 워크 플로우
> 데이터 → Tokenizer → DataCollator → Model → Trainner → 결과

1. 토크나이저
: 문장을 쪼개는 가위
- 사전학습때 사용된 토큰화 도구를 사용해서 입력 데이터 처리

2. 데이터 콜레이터
: 규격화된 통
- 입력 데이터를 일정한 크기로 맞춰줌(패딩 처리 등)

3. Trainner
: 관리자
- 데이터 전처리 ~ 학습, 평가, 저장까지 모든 과정을 한 번에 수행해주는 고수준 API

4. Training Arguments
- 학습 과정에서 필요한 하이퍼 파라미터들을 설정하는 클래스
- 학습률, 배치 크기, 에포크 등 학습 관련된 설정값을 관리

5. Downstream Task Classes
- fine tuning 대상 모델
- 사전 학습된 모델을 다양한 task에 맞게 활용할 수 있음

- 토큰화 / 인코딩

```python
# BERT용 토큰화 도구 가져오기
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained("monologg/kobert")


# 토큰화 함수 정의 (자:어떻게 변환할지 규칙 설정)
def tokenize_function(examples):
    # padding='max_length' : max_length를 padding 길이로 설정
    # truncation=True : max_length보다 크면 자름
    # max_length=128  : 문장의 최대 길이
    return tokenizer(examples["document"],# 영화 댓글 텍스트 컬럼
                     max_length=128,  # 모든 문장을 128 토큰으로 통일
                     padding='max_length',# 짧은 문장은 0으로 채워서 max_length(128)에 맞춤
                     truncation=True)# 128보다 긴 문장은 128에서 잘라냄
# 토큰화 수행
# 훈련 데이터 전체에 토큰화 함수 적용
train_dataset1 = train_dataset.map(tokenize_function,
                                   batched=True) # 한 번에 여러 문장씩 처리해서 속도 향상

# 검증 데이터 전체에 토큰화 함수 적용
val_dataset1 = val_dataset.map(tokenize_function,
                               batched=True)
```

- 모델 훈련

```python
# HuggingFace transformers에서 텍스트 분류용 모델 가져오기
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(
    "monologg/kobert", # 모델명
    num_labels=2 # 분류할 클래스 개수 (이진분류 - 긍정/부정)
)
```

```python
# HuggingFace의 학습 도구들 가져오기
# TrainingArguments = 학습 설정값 모음
# Trainer = 실제 학습을 실행하는 도구
from transformers import TrainingArguments, Trainer

# DataCollatorWithPadding:배치 내 패딩을 자동으로 처리해주는 도구
from transformers import DataCollatorWithPadding
```

```python
# DataCollatorWithPadding()
# 같은 배치(batch) 내에서 가장 긴 문장의 길이에 맞춰
# 나머지 짧은 문장들의 뒤에 패딩(padding) 토큰을 자동으로 채워주는 기능
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# 학습 설정값 정의
training_args = TrainingArguments(
    output_dir='./results/kobert_naver', # 학습된 모델 저장할 폴더
    learning_rate=2e-5,# 학습률 0.00002 # W값을 한 번에 얼마나 업데이트할지  # 너무 크면 학습 불안정, 너무 작으면 학습 느림
    weight_decay=0.01,# 학습률 감쇄 # 과적합 방지용 (W값이 너무 커지지 않게)
    per_device_train_batch_size=16, # 훈련 데이터 배치 사이즈 # 훈련할 때 한 번에 16개 문장씩 처리
    per_device_eval_batch_size=16, # 검증 데이터 배치 사이즈 # 검증할 때 한 번에 16개 문장씩 처리
    num_train_epochs=1,# 학습 반복 수 # 전체 데이터를 1번 반복 학습 (데이터 전체를 1바퀴 돔) # 자:16배치* 1,250번 = 20,000개 데이터 (1바퀴)
    eval_strategy="epoch",# 검증/저장 시기 # 1 epoch마다 검증 데이터로 성능 측정
    save_strategy="epoch",# 베스트 모델 저장 여부 # 1 epoch마다 모델 저장
    load_best_model_at_end=True, # 학습 끝나면 가장 성능 좋은 모델 불러오기
    report_to = "none" # 모니터링 도구 사용 안함
                        # wandb (인공지능이 학습할 때 점수(Loss)가 예쁘게 잘 떨어지고 있는지, 정확도(Accuracy)는 올라가고 있는지를 실시간으로 확인)같은 모니터링 도구 사용 안 함
)

# Trainer 설정
# 모델, 학습설정, 데이터를 한 곳에 묶어주는 것
trainer = Trainer(
    model=model, # 학습할 모델 (KoBERT)
    args=training_args, # 위에서 설정한 학습 설정값
    train_dataset=train_dataset1, # 훈련 데이터(자: 문제집)
    eval_dataset=val_dataset1, # 검증 데이터(자:모의고사 문제집)
    data_collator=data_collator # 배치 패딩 처리 도구(자:보조코치)
)

# 학습 시작!
# 위 설정대로 KoBERT를 네이버 영화 댓글로 파인튜닝
trainer.train()
```
