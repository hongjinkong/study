### Matplotlib

 - 데이터 시각화를 도와주는 라이브러리
 - 데이터 시각화란, 분산된 데이터를 도표나 차트로 정리하는 행위다.

```python
# Matplotlib 불러오기
# Matplotlib 안에 있는 pyplot 모듈 불러오기

# 1. import 키워드만 사용해서 불러오기
import matplotlib.pyplot as plt

# 2. from 과 import 키워드를 사용해서 불러오기
from matplotlib import pyplot as plt
```

```python
# 선 그래프 그려보기 

# 그래프 그릴 데이터

# 필수 데이터 : y축 데이터
y = [2,4,6]


# 선 그래프 생성
plt.plot(y)

# 그래프 출력 명령
plt.show()
```

```python
# 그래프 설정 -> 그래프를 출력(show)하기 전에 생성된 그래프에 대해 설정
# -> figure를 통해서 그래프에 대한 전반적인 설정
# figuer 안에 있는 매개변수를 통해서 환경설정
# figsize = (가로크기, 세로크기)
# facecolor = 테두리 배경색 이름
plt.figure(figsize=(3,2), facecolor = 'green')
# 그래프 생성
x= [1,2,3]
y= [4,5,6]
plt.plot(x,y)

# 그래프 출력
plt.show()
```

```python
# 그래프 설정
plt.figure(figsize=(3, 3))

# 그래프 생성
# 선 꾸며주기 -> line style 설정
# marker -> x축과 y축에 대응되는 지점을 표현
# line width -> 선의 긁기 설정
# color -> 선의 색상 변경
# marker color -> marker의 색상 변경
# markerfacecolor -> marker의 내부 색상 변경
# markeredgecolor -> marker의 테두리 색상 변경
# ms = 마커의 크기
# mew = 마커 테두리 두께
plt.plot(x,y, ls = ':', marker = 'o', lw = 1, c='r',
         markerfacecolor='b', markeredgecolor='y', ms = 5, mew = 1)


# 그래프 출력
```

```python
# 한 번에 여러 개의 그래프를 그려보자
# x는 0부터 7까지, y는 [5,2,3,6,8,7,5,7]
import numpy as np

# 첫 번째 그래프

# 데이터 준비 
y = [5,2,3,6,8,7,5,7]
x = np.arange(0,8)

# 그래프 설정
plt.figure(figsize=(5,3))

# 그래프 생성
plt.plot(x,y, ls = '--', color = 'red', marker = '*', ms = 10)

# 그래프 출력
plt.show()

# 두 번째 그래프
y = [5,2,3,6,8,7,5,7]
x = np.arange(0,8)
# 데이터 준비


# 그래프 설정
plt.figure(figsize=(5,3))

# 그래프 생성
plt.plot(x,y, ls = ':', color = 'green', marker = 'd', ms = 10, mfc = 'blue', mec = 'violet')

# 그래프 출력
plt.show()

```

```python
# 그래프 가로 출력
# -> subplot을 통해서 그래프 출력 순서를 선언


# 그래프 설정
plt.figure(figsize=(10, 3))
# 그래프 생성

# subplot 생성
# subplot(행, 열, 다음에 올 그래프의 순서)
# 1개의 행 2개의 열에 공간을 선언하고, 지금부터 그릴 그래프는 첫 번째 figure에 그린다.
plt.subplot(1, 2, 1)
plt.plot(x, y, ls = '--', c='green')

# 1개의 행 2개의 열에 공간을 선언하고, 지금부터 그릴 그래프는 두 번째 figure에 그린다.
plt.subplot(1,2,2)
plt.plot(x, y, ls = '--', c='red')

# 그래프 출력
plt.show()

```

```python

x = np.arange(0,8)
y = [5,2,3,6,8,7,5,7]

plt.figure(figsize=(5,4))

plt.plot(x,y)
# y축의 범위를 설정 
# y축의 범위를 제한 -> plt.ylim(-2, 12)
plt.ylim(-2, 12)
# x축의 범위를 설정 -> plt.xlim(-4, 10)
plt.xlim(-4, 10)

# 그래프 격자무늬 생성
plt.grid()

# 그래프 출력
plt.show()
```

```python
# 선 그래프를 한 캔버스 안에 세 줄 그리기 

# 데이터 준비
x = [1,2,3]
# 그래프 설정
y1 = [1,2,3]
y2 = [2,4,6]
y3 = [3,6,9]

# 그래프 생성
plt.figure(figsize=(4,3))
plt.plot(x, y1, label='y1')
plt.plot(x, y2, label='y2')
plt.plot(x, y3, label='y3')
# 범례 표시
# 각각의 plot함수에 대해서 label을 붙여줘야 한다. -> legend함수 사용
plt.legend()


# 그래프 이름 지어주기 -> title함수 사용

# 한글로 이름 지어주기 -> plt.rcParams['font.family']로 글꼴을 바꿔주어야 한다.
# matplotlib 안에 있는 글꼴이 한글을 표현할 수 없음
#  -> matplotlib의 글꼴 자체를 바꿔줘야 함

plt.rcParams ['font.family'] = 'Gulim'
plt.title('내 그래프')
# rcParams → Matplotlib의 환경설정(기본 설정)을 저장하는 딕셔너리
plt.show()
```
