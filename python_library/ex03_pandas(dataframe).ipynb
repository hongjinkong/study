### loc, iloc
- (index) location
- 행, 열에 모두 접근 가능한 인덱서
- loc : 행의 실제 이름이나, 열의 실제 이름으로 인덱싱, 슬라이싱 하는 방법
- iloc : 행의 인덱스 번호나, 열의 인덱스 번호를 통해서 인덱싱, 슬라이싱 하는 방법

```python
# loc : 인덱스 이름으로 접근
# 행은 어디부터 어디까지 가져오고, 열은 어디부터 어디까지 가져올건지 동시에 입력
# 행접근, 열접근 동시에 입력
# df.loc[행에대한 접근, 열에대한 접근]
df
# 서울부터 인천까지의 데이터만 가져오고 싶을 때
df.loc['서울':'인천', ]
```

```python
# 2020년 데이터만 가져오고 싶을 때
df.loc['서울':'인천', 2020]
```

```python
# 데이터프레임 형태로 반환하고 싶으면?
df.loc['서울':'인천', [2020]]
df.loc[:,[2020]]
```

```python
# iloc : 인덱스 번호로만 접근 가능
# df.iloc[행에대한 접근, 열에대한 접근]

# 서울부터 부산까지의 데이터만 가져오고 싶을 때
df.iloc [0:2,0:2]
```

```python
# 2020년 데이터만 가져오고 싶을 때
df.iloc[:,[0]]
```

### 실습 예제

```python
dict1 = {
    '메뉴명': [
        '아메리카노', '카페라떼', '바닐라라떼', '카푸치노', '카라멜마끼아또',
        '콜드브루', '돌체라떼', '녹차라떼', '초코라떼', '에스프레소'
    ],
    '가격': [
        3000, 4000, 4500, 4200, 5000,
        4500, 4800, 4300, 4500, 2500
    ],
    '판매량' : [
        120, 95, 80, 70, 65, 110, 
        60, 50, 45, 150
    ]
}
```

```python
# 1. 데이터 프레임 생성
menu = pd.DataFrame(dict1)
menu
```

```python
# 2. 매출액을 계산해서 새로운 열에 저장하시오.
# 매출액 = 가격 * 판매량
# 가격 컬럼과 판매량 컬럼 인덱싱 해서 곱해주고 매출액 컬럼에 추가
```

```python
menu['매출액'] = menu['가격'] * menu['판매량']
menu
```

```python
# set_index() : 인덱스 재설정
menu.set_index('메뉴명', inplace=True)
```

```python
menu
```

```python
# 3. 아메리카노와 초코라떼의 가격과 판매량만 가져오시오.
menu.loc[['아메리카노', '초코라떼'], ['가격', '판매량']]
menu.iloc[[0,8],[0,1]]
```

```python
menu['매출액'] = menu['가격'] * menu['판매량']
menu
```

```python
# 4. 매출액이 30만원 이상 40만원 이하인 데이터만 가져오시오.
menu[(menu['매출액'] >= 300000) & (menu['매출액'] <= 400000)]
```

```python
# 또는 전치 사용해서 구하기
```

```python
score
```

```python
# 과목별 합계를 데이터프레임에 추가하기
score["과목별 합계"] = score.sum(axis = 1)
```

```python
score
```

```python
# 과목별 평균을 계산해서 데이터프레임에 추가하기
# mean 함수
# 합계 컬럼 제외한 나머지 컬럼의 평균을 계산하는 함수
score["과목별 평균"] = score.loc[:,'1반':'4반'].mean(axis=1)
score
```

```python
# 전체 반에서 과목별 가장 높은 점수만 가져오기
max_score = score.loc[ : , :'4반' ].max(axis=1)
```

```python
max_score
```

```python
# 전체 반에서 과목별 가장 낮은 점수만 가져오기
score.loc[:,:'4반'].min(axis=1)
```

### 카테고리 생성

```python
# 나이 데이터
ages = [1,2,10,21,23,37,31,61,20,41,32,99]
```

```python
ages
# 나이 데이터를 범위를 나눠서 아동, 청소년, 청년, 중장년, 노년으로 구분하고 싶음
```

```python
# 범위
# 1살 ~ 12살 : 아동
# 13살 ~ 19살 : 청소년
# 20살 ~ 39살 : 청년
# 40살 ~ 64살 : 중장년
# 65살 ~ 99살 : 노년

# 경계값 설정
bins = [0,12,19,39,64,99]
```

```python
# 카테고리 이름
labels = ['아동', '청소년', '청년', '중장년', '노년']
```

```python
#cut()
pd.cut(ages, bins = bins, labels = labels)
```

```python
df_ages = pd.DataFrame(ages, columns=['age'])
df_ages
```

```python
# 각 나이에 맞게 카테고리 추가하기
df_ages['age_cat'] = 'cats'
```

```python
df_ages
```

```python
# 카테고리별 데이터 개수 확인하기
# value_counts()
df_ages['age_cat'].value_counts()
```
