### 배열의 인덱싱과 슬라이싱

```python
import numpy as np
```

```python
# 배열의 인덱싱, 슬라이싱
# 1차원 형태의 배열은 리스트처럼 인덱싱, 슬라이싱이 가능하다.
arr6 = np.array([1,2,3])
arr6
```

```python
# 1에 접근
print(arr6[0])
```

```python
# 3에 접근
print(arr6[2])
```

```python
# 2차원 배열 인덱싱
arr7 = np.array([[1,2,3]
                ,[4,5,6]])
arr7
```

```python
# 3에 접근
# 0번 요소에 접근 -> 2번 요소에 접근
print(arr7[0][2])
```

```python
# 2차원 배열에서 행렬로 3에 접근하면, 0번째 행, 2번째 열에 접근해야한다.
# 행렬로 접근 -> arr[행에대한 접근, 열에대한 접근]
```

```python
# 1부터 50까지 들어있는 배열 -> arange함수를 통해서 생성
arr8 = np.arange(1,51)
arr8
```

```python
# (5, 10)형태로 형태를 바꿔주기 -> reshape(5,10)
arr8.reshape(5,10)
```

```python
arr8 = arr8.reshape(5,10)
arr8
```

```python
# 슬라이싱 -> [시작인덱스 : 끝인덱스 + 1]
# [행에 대한 접근, 열에 대한 접근]
arr8[1:3,1:5]
```

```python
# 열에 대한 접근은 생략가능 : 생략한다면, 모든 열을 가져온다.
arr8[1:3]
```

```python
# 모든 행을 가지고 오고 싶을 때
arr8[:,1:5]
```

### boolean indexing
- 배열 안에서 조건이 맞는 데이터만 가져오는 방법
- 외부 데이터를 불러올 때 필터링 해주는 기능

```python
es_score = np.array([30, 40, 50, 60, 70 , 80])
es_score
```

```python
# 60점 이상인 데이터만 가져오고 싶다면

# 조건식
# 배열은 요소 하나 하나 다 연산 가능하기 때문에, 비교 연산자를 넣으면 boolean값(True, False)이 반환된다.
bool1 = es_score >= 60
print(bool1)
```

```python
es_score[bool1]
```

```python
# 나는 es_score에서 60점 이상인 것들만 가져오겠다.
# -> boolean indexing(불리언 색인)
```

```python
import numpy as np
```

```python
# boolean indexing 실습
# arr에서 짝수인 데이터를 판별해서 0인 값을 담아줘라
# 1. 배열 생성
list1 = [[6,1,6,8,7], [2,7,9,1,5]]

arr = np.array(list1)
arr
```

```python
# 2. 짝수인 데이터만 가져오기
bool2 = arr % 2 == 0
arr[bool2]
```

```python
# 3. 짝수인 데이터에 0 넣어주기
arr[bool2] = 0
arr
```
