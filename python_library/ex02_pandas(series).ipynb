```python
# pandas 불러오기
import pandas as pd
```

### Series 생성

```python
# Series는 1차원 형태의 배열로 이루어져 있음

# 리스트 기반으로 Series 생성

# 네 개의 지역의 인구수 데이터
# 서울: 9,668,465명
# 부산: 3,391,946명
# 인천: 2,942,828명
# 광주: 1,450,062명

population = pd.Series([9668465, 3391946, 2942828, 1450062], 
          index=['서울', '부산', '인천', '광주'])
population
```

```python
# 딕셔너리 기반으로 Series 생성
# 딕셔너리는 key값과 value값으로 이루어져 있음
# Series는 index값과 value값으로 이루어져 있음

# key -> index
# value -> value

# 네 개의 지역의 인구수 데이터
# 서울: 9,668,465명
# 부산: 3,391,946명
# 인천: 2,942,828명
# 광주: 1,450,062명

dict1 = {'서울': 9668465, '부산': 3391946, '인천': 2942828, '광주': 1450062}
pd.Series(dict1)
```

### Series 연산

```python
# ndarray처럼 Series도 정수 연산이 기능
# 인구수 데이터를 백만 단위로 바꿔주고 싶을때
population / 1000000
```

```python
# 본격적인 연산을 해보자!
# 음식 영양소 데이터
food_name = ['치킨', '닭갈비', '삼겹살', '공기밥', '닭가슴살', '아보카도', '우유']
food_kcal = [840, 585, 661, 310, 120, 320, 122]
```

```python
food_data = pd.Series(data = food_kcal, index = food_name)
food_data
```

```python
# Series의 이름 = "음식 칼로리"
# Seires의 인덱스 이름 = "칼로리"
food_data.name = "음식 칼로리"
food_data.index.name = "칼로리"


food_data
```

```python
# 데이터의 기술 통계량 구하기

# 1. 총 합계
food_data.sum()
```

```python
# 2. 전체 평균
#mean 함수
food_data.mean()
```

```python
# 3. 최대값, 최소값
print("최대값:", food_data.max())
print("최소값:", food_data.min())
```

```python
# 4. 요소의 개수
#count 함수
food_data.count()
```

```python
# 위에서 구한 기술통계량을 한 번에 확인해보자
# 데이터의 통계 확인
# describe 함수
food_data.describe().round(2)

```

### Series 인덱싱, 슬라이싱

```python
# Seires 인덱싱
food_data
```

```python
#치킨의 칼로리를 가져와보자.
food_data[0]
```

```python
# Series에서는 인덱스명을 통해 인덱싱 가능
food_data
```

```python
# 닭갈비부터 공기밥까지 가져오기
food_data[1:4]
```

```python
# Series를 슬라이싱 할 때 인덱스 명으로 슬라이싱하면 끝 번호도 포함
food_data['닭갈비':'공기밥']
```

```python
# 원하는 정보만 가져오기
# 닭갈비, 닭가슴살만 가져오기
# 리스트로 인덱스명 혹은 인덱스 번호를 담아서 원하는 정보만 꺼내올 수 있음
# 한번에 여러 인덱스명을 넘겨줄 수 있음 -> 리스트 형태로 묶어서 전달해줘야 함
food_data[['닭갈비', '닭가슴살']]
```

```python
# 여러 개의 인덱스 번호를 넘겨줄 수도 있음 -> 리스트 형태로 묶어서 전달
food_data [1,4]
```

### Boolean 인덱싱
- Boolean -> True/False
- 조건식에 맞는 데이터만 필터링해서 가져오는 방법

```python
# 칼로리가 400 이상인 데이터만 가져오고 싶다면?
food_data[food_data >= 400]
```

```python
# 칼로리가 600 이하인 데이터만 가져오고 싶다면?
food_data[food_data <= 600]
```

```python
# 칼로리가 200 이상이고 500 이하인 음식은?
food_data[(food_data >= 200) & (food_data <= 500)]
```

```python
# Series에서 논리 연산을 하기 위해서는 비트 연산자를 사용해야 함
```

```python
# 연산자의 우선순위 설정을 위해 소괄호 사용
```

```python
# 칼로리가 200미만인 데이터와 600이상인 음식만 가져오려면?
food_data[(food_data < 200) | (food_data > 600)]
```
