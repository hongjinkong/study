### 임의의 숫자로 채워진 배열 생성

```python
# 배열을 임의의 모양만큼 만들어주는 함수

# 0으로 이루어진 배열 생성
# np.zeros(배열의 모양, dtype = 자료형)

arr5 = np.zeros((2,4), dtype = 'int')
arr5
# 초기값 생성(빈 데이터 공간 만들기)
# True / False 값을 숫자 0과 1로 표현
# 예를 들어 0으로 채워진 배열을 만들고 나중에 값을 하나씩 넣기
# 0으로 채워진 배열을 만들고(불합격) 합격인 데이터만 1로 바꾸기
```

```python
# 1로 이루어진 배열 생성
# ones()
arr6 = np.ones((3,2), dtype = 'int')
arr6
```

```python
# 0, 1 이외의 값으로 채워주고 싶을때
# full()
# -> 원하는 숫자를 원하는 형태로 만들어주는 함수
# np.full((배열의 형태), 넣고자 하는 자료형)
arr7 = np.full((5,), 2.0)
arr7
```

```python
# 랜덤으로 값을 채워주는 함수
# random
# np.random.randint(시작값, 끝 값, 배열 모양 -> size = ())
# 끝 값은 포함하지 않는다 (슬라이싱과 같다)
arr8 = np.random.randint(1, 11, size = (2,2))
arr8
```

```python
# 1부터 50까지 들어있는 배열을 만들어보자.

# 1. 1부터 50까지 들어있는 리스트 생성
# 1-1. 빈 리스트 생성
list4 = []


# 1-2. 반복문과 range 사용
for i in range(1,51):
    list4.append(i)
print(list4)
```

```python
#2. 1번에서 만든 리스트를 배열로 변환
np.array(list4)
```

```python
# numpy에 원하는 숫자만큼 배열로 만들어주는 함수 존재
# arange
# arange(시작할 값, 끝 값, 증감량) -> 시작 값부터 끝 값의 전까지 이루어진 배열
arr5 = np.arange(1, 51)
arr5
```

```python
# 배열을 생성할 때, 타입을 지정
# -> 배열 안에 있는 요소의 자료형을 일치시켜줄 수 있다.
# -> 배열은 각기 다른 자료형을 요소로 받아들일 수 없다.
arr6 = np.array([1,2,3], dtype= 'float')
arr6
```

```python
# 생성할 때 잘못 만들었다면, astype 함수를 통해 자료형을 바꿔줄 수 있다.
# arr.astype('자료형')
arr7 = np.array(['1',2,3], dtype= 'float')
arr7
```

```python
# astype은 원본 데이터에 영향을 미치지는 않는다 
# -> 따로 변수에 담아주거나, 같은 변수에 재정의 해주어야 사용 가능하다
# 다른 변수에 담기
arr7_int = arr7.astype('int')
arr7_int
```

### 배열 연산

```python
# 배열의 산술연산 가능한 속성
list1 = [1,2,3]
list2 = [4,5,6]

list1 + list2
```

```python
# list1,2를 기반으로 배열로 만들어서 산술연산 하면 어떻게 되는가? 
arr1 = np.array(list1)
arr2 = np.array(list2)
print(arr1, arr2)
```

```python
arr1 + arr2
# 배열은 배열끼리 더하면 각 인덱스 번호에 맞는 요소끼리 연산된다.
```

```python
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
print(arr1 % arr2)
```

```python
# 비교 연산자
print(arr1 == arr2)
print(arr1 > arr2)
print(arr1 < arr2)
```

```python
# 배열과 정수랑 연산 O
# 리스트는 정수랑 연산 X
# 배열은 정수랑 연산이 가능하고, 각 요소별로 연산 후 array형태로 반환
arr1 * 2
```

```python
# 2차원 배열은 어떻게 연산이 되는지 확인
# 2차원 배열 -> 2,3
arr3 = np.array([[1,2,3]
                 ,[4,5,6]])
arr4 = arr3 * 3
print(arr3)
print(arr4)
```

```python
arr3 + arr4
```

```python
# 배열끼리의 연산은, '배열 모양만 맞다면' 같은 인덱스 번호끼리 연산이 된다.
# 2차원 배열 -> 2,3
arr3 = np.array([[1,2,3]
                 ,[4,5,6]])

# 2차원 배열 -> 2, 4
arr4 = np.array([[7,8,9,10]
                 ,[11,12,13,14]])

print(arr3)
print(arr4)
print(arr3 + arr4)
```
