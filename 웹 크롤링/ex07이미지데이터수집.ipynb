### 네이버 이미지 검색을 통해서 이미지 수집하기~
- 이미지 경로를 통해 파일 다운로드 방법 알기!

```python
import os
import time

from selenium import webdriver as wd
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

# 이미지 경로를 이용해 파일로 변환해주는 라이브러리
from urllib.request import urlretrieve
```

```python
#이미지를 저장할 폴더 생성
folder_path = 'imgs'
if not os.path.isdir(folder_path):
    os.mkdir(folder_path)
```

```python
url = 'https://search.naver.com/search.naver?ssc=tab.image.all&where=image&sm=tab_jum&query=%EB%A1%9C%EB%A6%AC+%EB%A7%A5%EA%B8%B8%EB%A1%9C%EC%9D%B4'
```

```python
driver = wd.Chrome()
```

```python
driver.get(url)
```

```python
#스크롤 2회만 내리기
for i in range(2):
    body = driver.find_element(By.CSS_SELECTOR, 'body')
    body.send_keys(Keys.END)
    time.sleep(0.5)
```

```python
# 이미지 태그 수집
# CSS 선택자 : #main_pack > section > div.api_subject_bx._fe_image_tab_grid_root.ani_fadein > div > div > div.image_tile._fe_image_tab_grid > div > div > div > div > img
img_tags = driver.find_elements(By.CSS_SELECTOR, '#main_pack > section > div.api_subject_bx._fe_image_tab_grid_root.ani_fadein > div > div > div.image_tile._fe_image_tab_grid > div > div > div > div > img')
len(img_tags)
```

```python
img_tags[0].get_attribute('src')
```

```python
#반복문 사용해서 이미지 저장하기
for i in range(len(img_tags)):
    urlretrieve(img_tags[i].get_attribute('src'), f'imgs/로리맥길로이_{i+1}.jpg')
    time.sleep(0.5)
```

```python
driver.quit()
```
