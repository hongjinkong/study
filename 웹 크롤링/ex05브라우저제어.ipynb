```python
# selenium 불러오기
# 브라우저 제어 라이브러리
from selenium import webdriver as wd
# 선텍자 활용하기 위한 라이브러리
from selenium.webdriver.common.by import By
# 키보드 활용하기 위한 라이브러리   
from selenium.webdriver.common.keys import Keys
```

```python
# 1. 브라우저 실행하기
driver = wd.Chrome()
```

```python
# 2. 원하는 사이트로 이동(req.get 유사)
driver.get("https://www.naver.com")
```

```python
# 3. 검색어를 입력(검색창의 css선택자를 활용)
# css 선택자 : #query
search_input = driver.find_element(By.CSS_SELECTOR, "#query")
search_input
```

## selenium을 활용한 요소 선택
- driver변수가 브라우저 그 자체이기 때문에 처음부터 html데이터를 객체의 형식으로 가짐(변환x)
- driver.fine_element(s)(By.CSS_SELECTOR,'선택자')
- BeatuifulSoup의 select, select_one과 같은 역할
- selenium 은 requests와 BeatifulSoup의 역할을 동시에 수행

```python
#검색어 입력하기
search_input.send_keys("홍명보")
```

```python
#키보드의 엔터(Enter)를 입력(Keys 사용)
search_input.send_keys(Keys.ENTER)
```

```python
dir(Keys)
```

```python
#뉴스탭 클릭하기
#CSS 선택자 : #lnb > div.lnb_group > div > div.lnb_nav_area._nav_area_root > div:nth-child(2) > a
news_tab = driver.find_element(By.CSS_SELECTOR, "#lnb > div.lnb_group > div > div.lnb_nav_area._nav_area_root > div:nth-child(2) > a")
news_tab.click()
```

```python
#추가 데이터 로딩을 위한 화면 스크롤
#화면을 담고 있는 최상위 태그인 body에 end키를 입력!

body_tag = driver.find_element(By.TAG_NAME, "body")
body_tag.send_keys(Keys.END)
```

```python
import time
import random

#총 10번 반복해서 데이터를 로딩(end키)
for i in range(10):
    #데이터 로딩
    body_tag.send_keys(Keys.END)
    time.sleep(random.uniform(0.5,1.0))
```

```python
# 뉴스 타이틀 수집하기
# css 선택자 : a[data-heatmap-target=".tit"]

title_tags = driver.find_elements(By.CSS_SELECTOR, "a[data-heatmap-target='.tit']")
len(title_tags)
```

```python
for title_tag in title_tags:
    print(title_tag.text)
```
