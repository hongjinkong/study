# 웹 크롤링/ 웹 스크래핑


- 웹페이지에서 필요한 데이터를 수집하는 작업
- 웹크롤링 : 웹페이지를 따라가며 여러 페이지를 탐색하고 수집하는 작업
- 웹 스크래핑 : 특정 웹페이지 안에서 필요한 데이터만 추출하는 작업
- 사람이 웹 브라우저에서 페이지를 열고 필요한 정보를 복사하느 과정을 코드로 자동화 환 것
- 크롤링과 스크래핑은 특정 언어에만 한정된 개념은 아니며 다양한 언어로 구현 가능
- 단순히 함수를 외우기보다, 전체적인 흐름을 이해하는게 중요!

  요청 -> 응답 -> html파일 확인 -> 원하는 태그(정보가 담긴 곳) 선택(css선택자 활용) -> 원하는 정보 추출

## http 통신의 기초 : 요청과 응답

- 클라이언트 : 요청을 보내는 주체
- 서버  : 요청을 받고 응답을 보내주는 주체
- 브라우저도 결국 url 을 통해 서버에 요청하고 서버가 html/css/js 등을 응답하는 방식으로 동작

```python
# 요청 라이브러리 불러오기
import requests as req
```

```python
res = req.get("https://www.naver.com/")
res
```

```python
res.text
```

```python
#res (응답 객체)에는 http 통신과 관련한 다양한 정보가 들어있다.
print("응답 객체 타입", type(res))
print("요청 url", res.url)
print("응답 상태 코드", res.status_code)
print("인코딩", res.encoding)
```

- 400번대 : 요청 실패(클라이언트 문제로 인해 실패) 
- 500번대 : 요청 실패(서버 문제로 인해 실패)

## string(문자열) 데이터인 res.text를 html 객체 데이터로 변환해보자~
- 목적 : req로 받아온 데이터는 string 데이터이므로 쉽게 파악이 어렵다! 쉽게 파악하기 위해서 
객체 형태로 변환 필요
- 사용 라이브러리 : BeautifulSoup4

```python
!pip show beautifulsoup4

```

```python
from bs4 import BeautifulSoup as bs
```

```python
# beautifulsoup 은 html 파서를 따로 필요로 한다.
# lxml, html5lib 등은 이런 문제를 해결하기 위해 필요한 모듈이다.

!pip show lxml
```

```python
html = bs(res.text, 'lxml')
html
```

```python
# type 검사
# bs4.BeautifulSoup은 html 문자열을 파싱하여 태그 객체 트리(dom)처럼 객체화된 형태
# 태그의 속성(attribute)는 key/value형태로 dict처럼 접근가능
print(type(res.text), type(html))
```

```python
#실습 : 네이버에서 광주 날씨 검색해서 bs4를 활용해 html 객체로 변환하자!
search_irl = "https://search.naver.com/search.naver"
params = {
    'query' : '광주 날씨'
}
res = req.get(search_irl, params=params)
html = bs(res.text, 'lxml')
html
```

```python
#현재 온도를 가져오자
# css 선택자 : div.temperature_text > strong
html.select_one('div.temperature_text > strong').text
```

### CSS 선택자 종류
1. 아이디 선택자  : id 값으로 요소를 선택(#)
2. 클래스 선택자 : class 값으로 요소를 선택(.)
3. 타입(태그) 선택자 : 요소의 이름으로 요소를 선택(p, span, div.. 등)
4. 자식 선택자 : '>' 로 요소를 선택
5. 자손 선택자 : ' ' 로 요소를 선택
6. 형제 선택자 : '~'로 요소를 선택
7. 인접형제 선택자 : '+'로 요소를 선택
8. 이외에 속성 선택자, 가상 선택자 등으 추가로 활용

### BeautifulSoup 요소를 선택하는 함수

### 1. 'select()'
- css 선택자를 이용해서 **여러 요소**를 선택

### 2. 'select_one()'
- css 선택자를 이용해서 **하나의 요소**를 선택

## BeautifulSoup 요소에서 사용하는 주요 속성

### 1. 텍스트 추출 : text
- 태그 내부의 모든 텍스트를 반환
- 공백을 정리하지 않는다;.

### 2. 속성 : .get(), []
- .get(속성명), [속성명] 의 형태로 사용
- .get 사용시 속성이 없다면 'none'를 반환한다.
- [] 사용시 속성이 없다면 error
