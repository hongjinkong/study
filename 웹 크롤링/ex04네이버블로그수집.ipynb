# 네이버 블로그 데이터 수집
- 네이버 블로그에 접근해서 글 제목과 내용을 가져온다!
- iframe 구조를 이해하고 크롤링을 해보자

```python
#라이브러리 불러오기
import requests as req
from bs4 import BeautifulSoup as bs
```

```python
#블로그 검색 주소 : https://search.naver.com/search.naver?ssc=tab.blog.all&sm=tab_jum&query=%EA%B4%91%EC%A3%BC+%EB%8F%99%EB%AA%85%EB%8F%99+%EB%A7%9B%EC%A7%91
search_url = 'https://search.naver.com/search.naver'
params = {
    'ssc': 'tab.blog.all',
    'sm': 'tab_jum',
    'query': '광주 동명동 맛집'
}
res = req.get(search_url, params=params)
res
```

```python
res.url
```

```python
html = bs(res.text, 'lxml')
html
```

```python
#블로그 링크 수집
#css 선택자 : div > div > div > div > div.sds-comps-vertical-layout.sds-comps-full-layout.QeLMmbLV5oAf3sys> div > div.sds-comps-vertical-layout.sds-comps-full-layout > a
a_tags = html.select('div > div > div > div > div.sds-comps-vertical-layout.sds-comps-full-layout.QeLMmbLV5oAf3sys> div > div.sds-comps-vertical-layout.sds-comps-full-layout > a')

len(a_tags)
```

```python
#링크 추출
url_list = []
for a in a_tags:
    url_list.append(a.get('href'))
    
len(url_list)
```

```python
# 블로그 1개 (0번 블로그) 접속해서 제목, 내용 가져오기

res = req.get(url_list[0])
res
```

```python
html = bs(res.text, 'lxml')
html
```

```python
# 블로그 타이틀 수집
#css 수집자: div.se-module.se-module-text.se-title-text > p > span
html.select_one('div.se-module.se-module-text.se-title-text > p > span').text
```

## iframe 태그
- inline frame의 약자
- 해당 웹 페이지 안에 다른 html을 불러와서 삽입할 수 있는 기능을 제공
- html5 등장 이후로는 사용이 급감
- 출처가 같다면 DOM 접근 가능
- iframe 속 html 문서에 requests로 접그하기 위해서 iframe html 문서의 주소가 필요

```python
#블로그 제목을 못가져옴
# 이유 : iframe
# iframe 으로 이동한 후에 제목과 내용을 가져온다
iframe_url = html.select_one('#mainFrame').get('src')
iframe_url
```

```python
#상대경로로 되어 있는 iframe주소를 절대경로 변환
real_url = f'https://blog.naver.com{iframe_url}'
real_url
```

```python
#iframe 페이지를 요청하여 실제 제목을 다시 가져오자
res = req.get(real_url)
html = bs(res.text, 'lxml')
html
```

```python
#블로그 제목 가져오기
#css 수집자: div.se-module.se-module-text.se-title-text > p > span
html.select_one('div.se-module.se-module-text.se-title-text > p > span').text
```

```python
#블로그 내용 가져오기
#css 선택자 : div.se-main-container
blog_content = html.select_one('div.se-main-container').text
```

```python
# 정규표현식을 통한 불필요한 글자 지우기!
import re
def preprocess_sentence_kr(w):
  w = w.strip()
  w = re.sub(r"[^0-9가-힣?.!,¿]+", " ", w)
  w = w.strip()
  return w
```

```python
content = preprocess_sentence_kr(blog_content)
content
```

```python
import time
```

```python
title_list = []
content_list = []

for url in url_list:
    #1. 블로그 주소를 통해 블로그 페이지 요청
    res = req.get(url)
    html = bs(res.text, 'lxml')
    
    #2. iframe주소 가져오기(real_url 복원)
    iframe_url = html.select_one('#mainFrame').get('src')
    real_url = f'https://blog.naver.com{iframe_url}'
    
    #3. iframe주소를 통해 블로그 페이지 요청
    real_res = req.get(real_url)
    real_html = bs(real_res.text, 'lxml')
    
    #4. 블로그 제목과 블로그 내용을 가져오기
    title = real_html.select_one('div.se-module.se-module-text.se-title-text > p > span').text
    content = preprocess_sentence_kr(real_html.select_one('div.se-main-container').text)
    
    title_list.append(title)
    content_list.append(content)
    
    time.sleep(0.5)
```

```python
display(title_list)
display(content_list)
```

```python
print(len(url_list), len(content_list), len(title_list))
```

```python
import pandas as pd
df = pd.DataFrame({
    '글 제목' : title_list,
    '글 내용' : content_list,
    '글 주소' : url_list
})

df.head(30)
```
