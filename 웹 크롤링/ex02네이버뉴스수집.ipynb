# 여러 페이지의 내용을 수집해야 할 경우
- 여러 페이지가 모여 있는 카테고리 등의 페이지에서 url주소 수집
- a태그의 href속성값을 활요!
- 반복문을 사용해서 각 주소에 접근하여 데이터를 크롤링

## 데이터 수집 순서
1. 네이버 뉴스 카테고리 페이지에서 url 주소 수집(리스트에 저장)
2. 반복문을 활용하여 주소 리스트의 각 뉴스 주소로 html문서를 요청(requests)
3. 각 뉴스 페이지의 html문서를 html객체로 변환(BeautifulSoup)
4. 변환된 문서에서 css선택자를 활용해 뉴스 제목과 내용을 수집해서 저장(리스트에 저장)

```python
#라이브러리 불러오기
import requests as req
from bs4 import BeautifulSoup as bs
```

```python
news_section_url = 'https://news.naver.com/section/101'
res = req.get(news_section_url)
html = bs(res.text, 'lxml')
html
```

```python
# a태그의 href속성을 통해서 각 뉴스 주소 가져오기!
#css 선택자 : a.sa_text_title 
news_list = html.select('a.sa_text_title')
len(news_list)
```

```python
#반목문을 사용해서 a태그의 href속성값만을 추출
href_list = []
for news in news_list:
    href = news.get('href')
    href_list.append(href)
    
href_list = [news.get('href') for news in news_list]
href_list
```

```python
news_url ='https://n.news.naver.com/mnews/article/366/0001177934'
res = req.get(news_url)
html = bs(res.text, 'lxml')
html

html.select_one('#title_area > span').text

print(html.select_one('#dic_area').text)
```

```python
#href_list에 저장된 url주소로 요청하여 가져온 html 문서로부터
#뉴스 제목과 뉴스 내용을 추출해서 리스트에 저장해보자!
title_list = []
content_list = []

for url in href_list:
    res = req.get(url)
    html = bs(res.text, 'lxml')
    title = html.select_one('#title_area > span').text
    content = html.select_one('#dic_area').text
    title_list.append(title)
    content_list.append(content)
    
title_list
content_list
```

```python
#df를 만들기 전에 list들의 길이 확인하는 습관 가지기
print(len(href_list), len(title_list), len(content_list))
```

```python
import pandas as pd

df = pd.DataFrame({
    "뉴스 제목" : title_list,
    "뉴스 내용" : content_list,
    "URL" : href_list
})
df.head(4)
```

```python
df.to_excel("네이버_경제_뉴스_데이터수집.xlsx")
```
