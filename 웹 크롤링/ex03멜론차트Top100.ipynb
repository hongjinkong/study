# 멜론차트 Top 100 데이터 수집

- 순위, 노래 제목, 가수 이름을 수집한다.
- 음악차트 데이터는 대표적인 트렌드 데이터로 다양한 분야에서 활용이 가능

## 실습 목표
1. http요청을 보낼 때, header를 함께 보내는 방법을 알아내기
2. css 선택자를 쉽게 찾는 방법 알기


```python
import requests as req
from bs4 import BeautifulSoup as bs
import pandas as pd
```

```python
url = 'https://www.melon.com/chart/'
res = req.get(url)
res
```

```python
#user-agent : Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/150.0.0.0 Safari/537.36
# 크롤러는 브라우저가 아니기 때문에 서버는 해당요청을 종종 차단한다
# header 정보를 추가하여, 마치 브라우저에서 요청하는 것처럼 서버를 속이는 것이 가능하다

headers = {
    "user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/150.0.0.0 Safari/537.36"
}
```

```python
#요청을 할 때 headers를 함께 전송
res = req.get(url, headers=headers)
res
```

```python
html = bs(res.text, 'lxml')
html
```

```python
#노래 제목 수정
#css 선택자 : td:nth-child(6) > div > div > div.ellipsis.rank01 > span > a
title_tags = html.select('td:nth-child(6) > div > div > div.ellipsis.rank01 > span > a')
title_tags
```

```python
#가수 이름 수집
# css 선택자  : td:nth-child(6) > div > div > div.ellipsis.rank02 > span
singer_tags = html.select('td:nth-child(6) > div > div > div.ellipsis.rank02 > span')
singer_tags
```

```python
# 순위 수집
# CSS 선택자 : td:nth-child(2) > div > span.rank
rank_tags = html.select('td:nth-child(2) > div > span.rank')
rank_tags
```

```python
#개수 검증 
print(len(title_tags), len(singer_tags), len(rank_tags))
```

```python
#가수 이름, 노래제목, 순위 문자열만 따로 추출
title_list = [title.text for title in title_tags]
singer_list = [singer.text for singer in singer_tags]
rank_list = [rank.text for rank in rank_tags]

for i in range(len(title_list)):
    title_list.append(title_tags[i].text)
    singer_list.append(singer_tags[i].text)
    rank_list.append(rank_tags[i].text)
   
display(title_list)
display(singer_list)
display(rank_list)
```

```python
#데이터프레임 생성
df = pd.DataFrame({
    "순위": rank_list,
    "노래 제목": title_list,
    "가수": singer_list
})
df.head(5)
```

```python
# csv로 저장
df.to_csv('2026년 7월 9일 멜론차트 Top100.csv', index=False, encoding='utf-8-sig')
```
