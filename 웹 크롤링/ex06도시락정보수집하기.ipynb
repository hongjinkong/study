## 도시락 데이터 수집하기!! (실습)
- 한솥도시락에서 도시락 메뉴 정보를 수집해보자
- 수집한 도시락의 정보를 xlsx 파일로 저장해보자

```python
#필요 라이브러리 불러오기
import requests as req
import pandas as pd
from selenium import webdriver as wd
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
import random
```

```python
url = "https://www.hsd.co.kr/menu/menu_list"
driver = wd.Chrome()
driver.get(url)
```

```python
#더보기 버튼을 클릭하여 모든 메뉴가 화면에 뜨도록 제어하기(실습)
#css 선택자 : a.c_05

more_btn = driver.find_element(By.CSS_SELECTOR, "a.c_05")

while True:
    if not more_btn.is_displayed():
        print("메뉴의 끝입니다.")
        break
    
    more_btn.click()
    time.sleep(random.uniform(0.5, 1.0))
'''
while True:
    try:
        more_btn.click()
        time.sleep(random.uniform(0.5, 1.0))
    except:
        print("메뉴의 끝입니다.")
        break
'''
    
```

```python
#도시락의 상품명이 담긴 요소를 수집

title_tags = driver.find_elements(By.CSS_SELECTOR, 'h4.h.fz_03')

for title_tag in title_tags:
    print(title_tag.text)
```

```python
# 가격 정보가 담긴 요소를 수집! (실습)
# CSS 선택자 : div > div.item-text > div > strong
price_tags = driver.find_elements(By.CSS_SELECTOR, 'div > div.item-text > div > strong')
len(price_tags)
```

```python
menu_items = driver.find_elements(By.CSS_SELECTOR, "li.item")

for item in menu_items:
    li_id = item.get_attribute("id")        # 예: "menuList_694"
    menu_id = li_id.split("_")[-1]           # 예: "694"
    time.sleep(random.uniform(0.5, 1.0))
```

```python
menu_name_list = []
menu_price_list = []

for title_tag in title_tags:
    menu_name_list.append(title_tag.text)

for price_tag in price_tags:
    menu_price_list.append(price_tag.text)

print(len(menu_name_list), len(menu_price_list))
```

```python
##############################resquests 사용################################
# selenium 과 requests 사용하여 도시락 정보 수집하기
import requests as req
from bs4 import BeautifulSoup as bs

#url이 담긴 요소를 가져오기
a_tags = driver.find_elements(By.CSS_SELECTOR, 'div > a.item-cont')

# onclick으로부터 url 정보 추출하기
# url 형태 : https://www.hsd.co.kr/menu/menu_view/700?cate1=5&cate2=67
onclick = a_tags[1].get_attribute('onclick')
onclick_split = onclick.split("'")


menu_id = onclick_split[5]
cate1 = onclick_split[1]
cate2 = onclick_split[3]


url = f'https://www.hsd.co.kr/menu/menu_view/{menu_id}?cate1={cate1}&cate2={cate2}'

url_list = []


for a_tag in a_tags:
    onclick = a_tags[1].get_attribute('onclick')
    onclick_split = onclick.split("'")
    
    menu_id = onclick_split[5]
    cate1 = onclick_split[1]
    cate2 = onclick_split[3]


    url = f'https://www.hsd.co.kr/menu/menu_view/{menu_id}?cate1={cate1}&cate2={cate2}'
    url_list.append(url)

res = req.get(url_list[0])
html = bs(res.text, 'lxml')
#메뉴 설명 가져오기!
# CSS 선택자 : #container > div.content.menu > section > div > div.view_cont > div.menu_detail > div.txt_wrap > p
html.select_one('#container > div.content.menu > section > div > div.view_cont > div.menu_detail > div.txt_wrap > p').text
```

```python
description_list = []

for url in url_list:
    res = req.get(url)
    html = bs(res.text, 'lxml')
    description = html.select_one('#container > div.content.menu > section > div > div.view_cont > div.menu_detail > div.txt_wrap > p').text
    description_list.append(description)
    time.sleep(0.5)

len(description_list)
```

```python
df = pd.DataFrame({
    '메뉴명' : menu_name_list,
    '메뉴 가격' : menu_price_list,
    '메뉴 상세내용' : description_list
    })
df.head(30)
```

```python
df.to_excel("한솥 도시락 메뉴 정리.xlsx")
```
