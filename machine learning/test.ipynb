### 문제 7
- (1) 정확도 : 전체 데이터 중에서 모델이 맞게 예측한 비율

- (2) 재현율 : 실제로 양성인 것들 중에서 모델이 맞다고 잘 찾아낸 비율

- (3) 정밀도 : 모델이 양성이라고 한 것들 중에서 진짜로 맞는 비율

```python
# 문제 5-1

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_iris

# (1) iris data 불러오기
iris_data = load_iris()

# (2) iris data를 문제와 답으로 분리
X = iris_data['data']
y = iris_data['target']

# (3) 분리된 문제와 답을 train data와 test data로 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=28
)

# (4) KNN분류기 모델 생성 (이웃 수는 3) 및 학습
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# (5) 모델 평가 진행
score = knn.score(X_test, y_test)
print(score)
```
