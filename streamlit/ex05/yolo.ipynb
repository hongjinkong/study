```python
# Yolo : 이미지 or 영상 에서 객체 탐지 딥러닝 모델
# OPenCV : 이미지 or 영상 처리 라이브러리
from ultralytics import YOLO
import cv2
```

```python
model = YOLO('yolo11m-pose.pt')
```

```python
# 컴퓨터에 연결된 기본 웹캠 실행 (OpenCV)
cap = cv2.VideoCapture(0)  #0(기본웹캠), 1,2 ..(여러개일경우에 변경)

#웹캠이 정상적으로 실행되었는지 확인 (cap.isOpened() -> boolean(true(정상)/false(열리지 않은 경우)))

if not cap.isOpened():
    print("웹캠을 사용할 수 없습니다")
else:
    print("웹캠을 실행합니다")
    print("종료하려면 q를 누르세요")

    # 웹캠 영상을 가져오기 위해(프레임 1개) 무한 반복
    while True:
        # 현재 화면(frame) 가져오기 (cap.read())
        # success : 프레임을 정상적으로 가져왔는지 여부 (true-정상, false-실패)
        # frame : 가져온 프레임(이미지)
        success, frame = cap.read()

        # success 여부에 따라 이미지를 더이상 가져오지 못한 경우 반복문 종료
        if not success:
            print("프레임을 가져올 수 없습니다")
            break

        # yolo pose 모델에 frame을 전달하여 사람의 관절 위치 탐지
        results = model(frame)

        annotated_image = results[0].plot()
        
        # q를 입력했을  경우 반복문 종료
        if cv2.waitKey(1) == ord("q"):
            break


# 웹캠 사용 종료 -> 카메라 자원을 운영체제에 반환
cap.release()

# OpenCV 통해서 생성된 모든 창 닫기
cv2.destroyAllWindows()
```
