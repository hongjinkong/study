### 함수
- 특정한 입력을 받아 정해진 작업을 수행하고 결과를 반환하는 것
- 자주 사용하는 기능을 모아놓고 필요할 때마다 호출해서 다시 사용하는 것

```python
#두 수와 연산자를 입력 받아서 원하는 연산을 수행하는 함수를 정의
def cal(num1, num2, op):
    if op == "+":
        result = num1 + num2
    elif op == "-":
        result = num1 - num2
    else:
        print("연산자 오류")
    return result

num1 = int(input("첫번째 정수 입력: "))
num2 = int(input("두번째 정수 입력: "))
op = input("연산자 입력(+,-): ")
result = cal(num1, num2, op)
print("결과 : {}".format(result))
```

```python
#어떤 수를 거듭제곱해주는 함수 정의

def power(num, power):
    result = num ** power
    return result
 
num = int(input("정수 입력 : "))
exp = int(input("제곱 입력 : "))

print(f"{num}의 {exp}제곰은 {power(num,exp)}입니다.")
```

### 다수의 반환값 처리

```python
def add_sub(num1, num2):
    return num1+num2, num1 - num2
```

```python
add_sub(14,3)
```

```python
def sum_avg(*args):
    result_sum = sum(args)
    result_avg = result_sum / len(args)
    return result_avg, result_sum

result_avg, result_sum = sum_avg(10, 20, 30, 50)
print(sum_avg(10,20,30,50))
print(result_sum)  
print(result_avg)  
```
