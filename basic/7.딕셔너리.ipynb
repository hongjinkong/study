```python
#python 과목 평균 구하기

score_dict = {'이름':['민주','원희','윤아','모카','이로하'],
        'Python':[90,100,85,90,80],
        'Java':[85,80,100,95,85],
        'html/css':[75,70,90,80,90]}

py_avg = sum(score_dict['Python']) / len(score_dict['Python'])

print("Python 과목 평균: ", py_avg)
```

```python
#각 과목 평균 구하
for subject, scores in score_dict.items():
    if subject == '이름':  
        continue
    
    avg = sum(scores) / len(scores)
    print(f"{subject} 과목 평균: {avg}")
```

```python
#영어 단어장을 만들고 사용자가 뜻을 검색하면 출력하세요
word_dict = {'apple': '사과', 'banana': '바나나', 'computer': '컴퓨터', 'python': '파이썬'}
a = str(input("뜻을 알고 싶은 영어 단어를 입력하세요 : "))

if a not in word_dict:
    print("등록되지 않은 단어입니다.")
else:
    print(word_dict.get(a))
```
