### 리스트 인덱싱과 슬라이싱
- 리스트에 들어있는 요소들에 접근하는 방법 -> 문자열에서의 방법과 동일
- 1. 인덱싱:원하는 값 하나를 가져온다는 의미
- 2. 슬라이싱: 원하는 구간의 값들을 가져온다는 의미

```python
list = [2, 5, 7, 9, 10]

list[0]

list[3]

list[-2]
```

```python
list2 = [1, 2, 3, ["a", "b", "c"]]
list2
list2[3]
```

```python
list2[3][1]
```

```python
#슬라이싱
list3 = [0,1,2,3,4]
list3[1:3]
list3[3:]
```

### 튜플
- 순서가 있는 객체의 집합
- 소괄호'()'로 작성, 내부의 값은 콤마(,)로 구분
- 하나의 튜플에 다양한 자료형 포함 가능
- 추가, 수정, 삭제 불가능

```python
# 튜플 선언 연습

# 튜플 선언하는 거 리스트와 전부 동일합니다
# 딱 하나 다른 건 대괄호가 아니라 소괄호 사용한다는 거만 유의해주시면 되겠습니다

# 1. 빈 튜플
tuple1 = ()

# 2. 숫자형 요소만 포함하는 튜플
tuple2 = (1, 2, 3)

# 3. 문자열 요소만 포함하는 튜플
tuple3 = ("My", "name")

# 4. 숫자형 + 문자열 요소를 포함하는 튜플
tuple4 = (1, 2, "My", "name")

# 5. 튜플 안에 튜플도 가능
tuple5 = (1, 2, ("My", "name"))
```

### 튜플인덱싱과 슬라이싱
- 인덱싱 : 원하는 값 하나를 가져온다
- 슬라이싱 : 원하는 구간을 가져온다

```python
# 인덱싱
tuple1 = (0,1,2,3,("a,","b","c"),5)

#1. 2를 가져오기
tuple1[2]
#2. ("a","b","c")가져오기
tuple1[4]
#3. 'b'를 가져오졍
tuple1[4][1]
```

```python
#슬라이싱ㅣ
#1. (1,2)를 가져오기
tuple1[1:3]
#2. (3,('a','b','c'),5) 가져왹
tuple1[3:]
```

```python
ramen = ["신라면", "진라면", "짜파게티", "너구리", "불닭볶음면", "안성탕면", "육개장", "팔도비빔면", "열라면", "삼양라면"]

name = str(input("라면입력 : "))


if name in ramen:
    print(f"{ramen.index(name)}번째 인덱스에 있습니다.")
else:
    ramen.append(name)
    print(ramen)
```
