### 반복문
- 똑같은 실행문장을 반복해서 수행할 떄 사용하는 문법
- while/ for

```python
# 1 부터 10까지 출력
num = 1
print(num)

while num < 11:

    num += 1
    print(num)
```

```python
while True:
    print("무한루프")
    break
```

```python
#운동을 1세트 끝날때마다 1세트완료 출력단 5세트 완료뒤에는 운동종료 출력

workout=0

while True:
    workout+=1
    print(f"{workout}세트 완료~")
    
    if workout == 5:
        print("운동종료!")
        break
```

```python
#몬스터 사냥 게임
#1. 체력이 100인 몬스터를 공격할때마다 랜덤 데미지 (5부터 30사이)
#2. 체력이 0 이하가 되면 종료

import random

hp = 100
attack_count = 0

while hp>0:
    damage = random.randint(1,9)
    hp -= damage
    attack_count += 1 
    
    if hp < 0:
        hp = 0
    
    print(f"{damage}데미지! 남은 체력 :{hp}")
       
print(f"몬스터 처치!(총 {attack_count}번 공격)")
```

```python
#1. 0부터 100 사이 중 3의 배수 합
sum=0

for i in range(100):
    if(i % 3 == 0):
        sum+=i
    
print("3의배수들의 합: ",  sum)
```

```python
#2. for 문을 이용하여 구구단 2단을 출력

for i in range(9):
    print(f"2 * {i+1} = {2*(i+1)}")
```

```python
# "물을 한잔 마셨습니다."를 ㄹ력출ㄹ하다가 10잔을 마시면 멈추기

#1. while문
i=0
while True:
    i += 1
    print(f"물을 {i}잔 마셨습니다.")
    if i == 10:
        print("물을 충분히 마셨습니다.이제그만!")
        break
```

```python
#2. for문

i=0
for i in range(10):
    i += 1
    print(f"물을 {i}잔 마셨습니다.")

print("물을 충분히 마셨습니다. 이제 그만!")
```

```python
#랜덤으로 1부터 50사이의 숫자를 뽑으면 뽑은 숫자를 맞추는  up,down게임
import random

answer = random.randint(1, 50)

while True:
    i = int(input("정수 입력: "))
    if i > answer:
        print(f"{i}보다 작은 수 입니다.")
    elif i < answer:
        print(f"{i}보다 큰 수 입니다.")
    else: 
        print(f"정답! 정답은 {answer}입니다.")
        break
```
