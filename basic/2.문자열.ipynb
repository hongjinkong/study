###문자열 포매팅

```python
day = 25
print("오늘은 6월", day, "일입니다.")
```

```python
date = 23
weather = "흐림"
```

```python
#print()
#1. format() 함수 이용
print("오늘은 6월 {}일입니다.".format(date))
print("오늘의 날씨는 {}입니다.".format(weather))
```

```python
date = 23
weather = "흐림"
month = 6
print("오늘은 {}월 {}일입니다. {}의 날씨는 {}입니다.".format(month, date, month, weather))
```

```python
#f-string 이용
print(f"오늘은 {month}월 {date}일입니다. {month}월 {date}일의 날씨는 {weather}입니다.")
```

```python
#3. %기호를 활용한 포매팅

print("오늘은 %d월 %d일입니다. %d일의 날씨는 %s입니다." %(month, date, date, weather))
```

```python
x = 100
y = 200
sum2 = x + y
print("%d와 %d의 합은 %d입니다." %(x, y, sum2))
print("{}와 {}의 합은 {}입니다.".format(x, y, sum2))
print(f"{x}와 {y}의 합은 {sum2}입니다.")
```

### 문자열 함수


```python
#coubnt() 함수 이용
a = "안녕하세요요요요요"
a.count("요")
```

```python
#find() 함수 이용
b = "Hello"
b.find("o")
```

```python
#index() 함수 이용
#index는 예외를 발생시키므로, 찾는 문자가 없을 때 오류가 발생한다외
b.index("o")
```

```python
#join() 함수 이용 -> 문자열 사이사이에 내가 넣고싶은 문자 추가
c = "홍진성"
"*".join(c)
```

```python
d = "like"
d.upper()
```

```python
#strip() 함수 이용 -> 문자열 양쪽에 있는 공백 제거
e = "   Hello World!   "
e.strip()
```

```python
#replace() 함수 이용 -> 문자열에서 특정 문자를 다른 문자로 바꿔주는 함수
f = "Hello World!"
f.replace("Hello", "Hi")
```

```python
#split() 함수 이용 -> 문자열을 특정 구분자로 나누어 리스트로 반환하는 함수
g = "apple,banana,orange"
g.split(",")
```
