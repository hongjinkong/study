### 연산자


```python
num1= 10
num2 = 7 
#곱하기
print(num1 * num2)
#나누기
print(num1 / num2)
#나머지
print(num1 % num2)
#몫
print(num1 // num2)
#제곱
print(num1 ** num2)
```

```python
#문자열 더하기
str1 = "Hello"
str2 = "World"
print(str1 + str2)
```

```python
#문자열과 숫자 더하기
str3 = "10"
print(num2 + int(str3)) #문자열을 숫자로 바꿔서 더하기
print(str(num1) + str3) #숫자를 문자열로 바꿔서 더하기
type(num1) #숫자형
```

```python
num1 = 23
num2 =3

print("더하기 결과: ", num1 + num2)
print("빼기 결과: ", num1 - num2)
print("곱하기 결과: ", num1 * num2)
print("나누기 결과: ", f"{num1 / num2:.2f}")
```

```python
num = int(input("숫자를 입력하세요: "))
type(num) #입력받은 값은 문자열형
```

```python
a = int(input("python 점수 입력 >> "))
b = int(input("머신러닝 점수 입력 >> "))
c = int(input("딥러닝 점수 입력 >> "))
total = a + b + c
avg = total / 3
print("합계 :{} ".format(total))
print("평균 :{} ".format(avg))
```

```python
num = 2
power = 10
print(num ** power)

```

```python
num = int(input("정수 입력 >> "))
power = int(input("제곱 입력 >> "))
print("{}의 {}승은 {}입니다.".format(num, power, num ** power))
```

### 비교연산자
- > <  <= >= ==!=

```python
a = 3
b = 7
print(a > b) #False
print(a < b) #True
print(a == b) #False
print(a != b) #True
```

### 논리연산자
- not, and, or

```python
#not
a = 3
b = 7
print(not a < b)
print(not a == b)
```

```python
#and
3 > 5 and 3 < 5
```

```python
time = int(input("시간>>"))
hour = time // 3600
minute = (time % 3600) // 60
second = time % 60
print("{}시간 {}분 {}초".format(hour, minute, second))
```
