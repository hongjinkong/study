```python
score = int(input("점수 입력: "))

if score>=90:
    grade = "A"
elif score>=80:
    grade = "B" 
elif score>=70:
    grade = "C"
elif score>=60:
    grade = "D"
else:
    grade = "F"
print(f"{score}점은 {grade}학점입니다.") 
```
