```python
# 문제 3-1
name = input("이름: ")
number = input("주민번호 7자리까지 입력: ")

year = number[:2]
month = number[2:4]
day = number[4:6]
sex = number[6:]

print("년도 ", year)
print("월 ", month)
print("일 ", day)
print("성별 ", sex)
```

```python
# 문제 3-2
if sex == '1' or '3':
    sex = "남자"
else: 
    sex = "여자"
    
print(f"{name}님은 {year}년 {month}월 {day}일에 태어났고, 성별은 {sex}입니다.")
```
