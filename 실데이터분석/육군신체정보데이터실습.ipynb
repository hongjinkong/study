```python
import pandas as pd
import numpy as np
```

```python
# 🎈 연습- 육군신체정보데이터 실습
data = pd.read_csv("data/육군신체측정정보.csv")
data
```

```python
#bmi 지수 구하기
#데이터확인하기
#확인해야하는 것 -> 결측치가 존재하는가, 각 컬럼의 Dtype이 어떻게 구성되어있는가?
data.info()

#문제점1 - bmi 지수를 구하기 위해서는 숫자 데이터의 연산이 필요한데 object임
#object -> float 형변화
#문제점2 - 결측치 존재
#결측치를 채워주거나 결측치가 존재하는 행을 삭제
```

```python
#특정 문자를 대체해줄 수 있는 문자열 함수 -> replace()
#cm. kg이라는 문자열을 공백으로 대체
height = data["신장 센티미터"].str.replace("cm", "")
height = height.astype(float)
height_meter = height * 0.01

weight = data["몸무게 킬로그램"].str.replace("kg", "")
weight = weight.astype(float)
```

```python
bmi = weight / (height_meter ** 2)
data['BMI'] = bmi
```

```python
# 정상체중인 사람 조회
# 'bmi' 컬럼이 18.5 이상 22.9 미만인 행만 조회
normal_weight = data[(data['BMI'] >= 18.5) & (data['BMI'] < 22.9)]

display(normal_weight)
```

```python
#결측치 전처리
#결측치를 확인할 수 있는 남수 -> isnull()
data.isnull().sum()
```

```python
#가슴둘레센티미터의 결측치가 존재하는데 그 행을 가져와보자
data[data["가슴 둘레 센티미터"].isnull()]
```

### 결측치 처리하는 방법
- 평균값으로 채우는 방법(분산이 작을 경우)
- 중앙값으로 채우는 방법(분산이 클 경우)
- 최빈값으로 채우는 방법(문자 등의 데이터)
- 삭제하는 방법

```python
#가슴둘레센티미터의 평균값을 구하려고함!
#모든 가슴둘레 센티미터 데이터가 cm가 붙어있는게 아니다.
#float 형태로 되어있는 데이터라면 그냥 그대로 쓰고, cm가 붙어있따면 공백으로 대체
def cm_replace(data):
    if type(data) == float:
        return data
    else:
        return data.replace(" cm", "")
    
```

```python
#1. apply 함수를 통해 cm_replace함수 활용하기
data["가슴 둘레 센티미터"] = data["가슴 둘레 센티미터"].apply(cm_replace).astype(float)
```

```python
# 2. 평균값 구하기
chest_avg = data["가슴 둘레 센티미터"].mean()
```

```python
data[data["가슴 둘레 센티미터"].isnull()]
```

```python
#결측치를 가슴 둘레 평균으로 대체하자!
data.loc[data["가슴 둘레 센티미터"].isnull(), "가슴 둘레 센티미터"] = chest_avg
```
