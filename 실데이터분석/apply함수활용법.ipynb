```python
import pandas as pd

score = pd.read_csv('data/score.csv', index_col="과목")
score
```

```python
#최대값에서 최소값을 빼는 함수를 정의해서 데이터프레임에 적용
def max_min(x):
    return x.max() - x.min()
```

```python
#데이터프레임에 사용자정의함수를 적용하는 방법 -> apply 
score.apply(max_min, axis=0) # 행끼리 비교
score.apply(max_min, axis=1) # 열끼리 비교
```

```python
date_df = pd.DataFrame({"yyyy-mm-dd":['2003-03-13','2017-10-20','2022-05-28']},
            columns = ['yyyy-mm-dd'])
date_df 
```

```python
#연도만을 가져올 수 있는 사용자 점의 함수 정의하기
def extract_year(x):
    return x.split('-')[0]
```

```python
date_df['tear'] = date_df['yyyy-mm-dd'].apply(extract_year)
```
